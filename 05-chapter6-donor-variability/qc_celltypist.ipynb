```python
donor_6 = sc.read('donor_6.h5ad')
donor_15 = sc.read('donor_15.h5ad')
donor_64 = sc.read('donor_64.h5ad')
donor_90 = sc.read('donor_90.h5ad')
```

```python
donor_6.obs['sample'] = 'Donor_6'
donor_6.obs['l1_capacity'] = 'low'

donor_15.obs['sample'] = 'Donor_15'
donor_15.obs['l1_capacity'] = 'low'

donor_64.obs['sample'] = 'Donor_64'
donor_64.obs['l1_capacity'] = 'high'

donor_90.obs['sample'] = 'Donor_90'
donor_90.obs['l1_capacity'] = 'high'
```

```python
# mitochondrial genes, "MT-" for human, "Mt-" for mouse
donor_6.var["mt"] = donor_6.var_names.str.startswith("MT-")
# ribosomal genes
donor_6.var["ribo"] = donor_6.var_names.str.startswith(("RPS", "RPL"))
# hemoglobin genes
donor_6.var["hb"] = donor_6.var_names.str.contains("^HB[^(P)]")
sc.pp.calculate_qc_metrics(
    donor_6, qc_vars=["mt", "ribo", "hb"], inplace=True, log1p=True
)
```

```python
# mitochondrial genes, "MT-" for human, "Mt-" for mouse
donor_15.var["mt"] = donor_15.var_names.str.startswith("MT-")
# ribosomal genes
donor_15.var["ribo"] = donor_15.var_names.str.startswith(("RPS", "RPL"))
# hemoglobin genes
donor_15.var["hb"] = donor_15.var_names.str.contains("^HB[^(P)]")
sc.pp.calculate_qc_metrics(
    donor_15, qc_vars=["mt", "ribo", "hb"], inplace=True, log1p=True
)
```

```python
# mitochondrial genes, "MT-" for human, "Mt-" for mouse
donor_64.var["mt"] = donor_64.var_names.str.startswith("MT-")
# ribosomal genes
donor_64.var["ribo"] = donor_64.var_names.str.startswith(("RPS", "RPL"))
# hemoglobin genes
donor_64.var["hb"] = donor_64.var_names.str.contains("^HB[^(P)]")
sc.pp.calculate_qc_metrics(
    donor_64, qc_vars=["mt", "ribo", "hb"], inplace=True, log1p=True
)
```

```python
# mitochondrial genes, "MT-" for human, "Mt-" for mouse
donor_90.var["mt"] = donor_90.var_names.str.startswith("MT-")
# ribosomal genes
donor_90.var["ribo"] = donor_90.var_names.str.startswith(("RPS", "RPL"))
# hemoglobin genes
donor_90.var["hb"] = donor_90.var_names.str.contains("^HB[^(P)]")
sc.pp.calculate_qc_metrics(
    donor_90, qc_vars=["mt", "ribo", "hb"], inplace=True, log1p=True
)
```
